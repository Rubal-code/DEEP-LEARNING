```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(-1,1,100)
x[:10]
```

```python
# define the quantization parameters
min_x = -1.0
max_x = 1.0
qmin = -128
qmax = 127

# calculate scale
scale = (max_x - min_x) / (qmax - qmin)

scale
```

```python
zero_point = 0
#quantization
q = np.round((x - zero_point) / scale).astype(int)

q=np.clip(q,qmin,qmax)

q[:10]
```

```python
# dequantization
x_hat=scale * (q - zero_point)
x_hat[:10]
```

#(NF4)Normal float 4 quantization

```python
#step 1 define the 16 nf4 values
nf4_values=np.array([
    -1,-0.75,-0.5,-0.35,-0.2,-0.1,-0.05,-0.01,
    0.01,0.05,0.1,0.2,0.35,0.5,0.75,1
])
# simulate the original float values
x=np.linspace(-1,1,100)
x[:10]
```

```python
# step 3 : quantize the x by mapping to nearest nf4 value
def quantize_to_nf4(x_vals,nf4_luft):
  indices=np.argmin(np.abs(x_vals[:,None] - nf4_luft[None,:]),axis=1)
  return nf4_luft[indices]

q=quantize_to_nf4(x,nf4_values)
q[:10]
```
