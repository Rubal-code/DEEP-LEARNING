```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

x=np.linspace(-1,1,100)
x[:10]
```

```python
# define the quantization parameters
min_x = -1.0
max_x = 1.0
qmin = -128
qmax = 127

# calculate scale
scale = (max_x - min_x) / (qmax - qmin)

scale
```

```python
zero_point = 0
#quantization
q = np.round((x - zero_point) / scale).astype(int)

q=np.clip(q,qmin,qmax)

q[:10]
```

```python
# dequantization
x_hat=scale * (q - zero_point)
x_hat[:10]
```
