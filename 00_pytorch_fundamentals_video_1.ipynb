##Random tensors

why random tensors?

Random tensors are important because the way many neural networks learn is that they start with the random numbers and then adjust those random numbers to better represent the number

`Start with random numbers -> look at the data -> update random numbers -> look at random -> process goes on...`

```python
# creating a random tensor
random_tensor = torch.rand(1,3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# create a random tensor with similar shape to the image tensor
random_image_size_tensor = torch.rand(size=(224,224,3)) # height, width, colour channels
random_image_size_tensor.shape, random_image_size_tensor.ndim
```
