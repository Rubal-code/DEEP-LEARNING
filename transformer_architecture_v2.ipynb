```python
# ============================================================
# PART 1: IMPORTS
# ============================================================

import math

import torch
import torch.nn as nn
import torch.nn.functional as F
```

```python
class PositionalEncoding(nn.Module):

    def __init__(self, d_model, max_len=5000):

        super().__init__()

        # Create matrix of shape:
        # (max_len, d_model)
        pe = torch.zeros(max_len, d_model)

        # Position:
        # [0, 1, 2, 3, ..., max_len-1]
        position = torch.arange(
            0,
            max_len,
            dtype=torch.float
        ).unsqueeze(1)

        # Different frequencies for each dimension
        div_term = torch.exp(
            torch.arange(0, d_model, 2).float()
            * (-math.log(10000.0) / d_model)
        )

        # sin for even dimensions
        pe[:, 0::2] = torch.sin(
            position * div_term
        )

        # cos for odd dimensions
        pe[:, 1::2] = torch.cos(
            position * div_term
        )

        # Add batch dimension
        # (1, max_len, d_model)
        pe = pe.unsqueeze(0)

        self.register_buffer("pe", pe)

    def forward(self, x):

        # x:
        # (batch_size, seq_len, d_model)

        seq_len = x.size(1)

        return x + self.pe[:, :seq_len]
```
