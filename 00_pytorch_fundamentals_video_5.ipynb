##Random tensors

why random tensors?

Random tensors are important because the way many neural networks learn is that they start with the random numbers and then adjust those random numbers to better represent the number

`Start with random numbers -> look at the data -> update random numbers -> look at random -> process goes on...`

```python
# creating a random tensor
random_tensor = torch.rand(1,3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python
# create a random tensor with similar shape to the image tensor
random_image_size_tensor = torch.rand(size=(224,224,3)) # height, width, colour channels
random_image_size_tensor.shape, random_image_size_tensor.ndim
```

#Tensor datatypes
**Note:** Tensor datatypes is one of the 3 big errors you will run into with deep learning and pytorch:

1.Tensors are not datatype

2.Tensors are not right shape

3.Tensors are not on right device

```python
# float32 tensor
float_32_tensor = torch.tensor([3.0, 6.0, 9.0],
                               dtype=None, # what datatype is the tensor (e.g. float32 or float16)
                               device=None, # what device is your tensor on
                               requires_grad=False) # whether or not to track gradients with this tensors operations
float_32_tensor
```

```python
float_32_tensor.dtype
```

```python
float_16_tensor = float_32_tensor.type(torch.float16)
float_16_tensor
```

```python
float_32_tensor * float_16_tensor
```

```python
int_32_tensor = torch.tensor([3, 6, 9], dtype=torch.int32)
int_32_tensor
```

```python
float_32_tensor * int_32_tensor
```

#getting tensors information

1. Tensors not right datatype - to do get datatype from a tensor , can use `tensor.dtype`

2. Tensors are not right shape - to get shape from a tensor , can use the `tensor.shape`

3. Tensors are not on the right device - to get device from the tensor , can use the `tensor.device`

```python
# create a tensor
some_tensor = torch.rand(3,4)
some_tensor
```

```python
# find out the details of the tensor
print(some_tensor)
print("the shape of the tensor",{some_tensor.shape})
print("the datatype of the tensor",{some_tensor.dtype})
print("the device of the tensor",{some_tensor.device})
```

# matrix multiplication

Two main ways of performing multiplication in neural networks and deep learning:

1. Element - wise multiplication

2. Matrix multiplication (dot wise product)

```python
# element wise multiplication
print(tensor,"*",tensor)
print(f"Equals: {tensor * tensor}")
```

```python
# matrix multiplication
torch.matmul(tensor,tensor)
```

```python
%%time
value = 0
for i in range(len(tensor)):
  value += tensor[i] * tensor[i]
print(value)
```

```python
%%time
torch.matmul(tensor,tensor)
```
